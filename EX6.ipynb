```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("EX6.jpg")
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
plt.imshow(gray, cmap='gray')
plt.title("Original Gray Image")
plt.axis("off")
plt.show()
```

```python
#sobel
sx = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3)
sy = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=3)

sobel = cv2.magnitude(sx, sy)

plt.imshow(sobel, cmap='gray')
plt.title("Sobel Edge Detection")
plt.axis("off")
plt.show()
```

```python
#prewett
px = np.array([[-1, 0, 1],
               [-1, 0, 1],
               [-1, 0, 1]])

py = np.array([[-1, -1, -1],
               [0, 0, 0],
               [1, 1, 1]])

prewitt_x = cv2.filter2D(gray, -1, px)
prewitt_y = cv2.filter2D(gray, -1, py)

prewitt = cv2.magnitude(prewitt_x.astype(np.float32),
                        prewitt_y.astype(np.float32))

plt.imshow(prewitt, cmap='gray')
plt.title("Prewitt Edge Detection")
plt.axis("off")
plt.show()
```

```python
#roberts
rx = np.array([[1, 0],
               [0, -1]])

ry = np.array([[0, 1],
               [-1, 0]])

roberts_x = cv2.filter2D(gray, -1, rx)
roberts_y = cv2.filter2D(gray, -1, ry)

roberts = cv2.magnitude(roberts_x.astype(np.float32),
                        roberts_y.astype(np.float32))

plt.imshow(roberts, cmap='gray')
plt.title("Roberts Edge Detection")
plt.axis("off")
plt.show()
```

```python
#laplacian
laplacian = cv2.Laplacian(gray, cv2.CV_64F)

plt.imshow(laplacian, cmap='gray')
plt.title("Laplacian Edge Detection")
plt.axis("off")
plt.show()
```

```python
#canny
canny = cv2.Canny(gray, 100, 200)

plt.imshow(canny, cmap='gray')
plt.title("Canny Edge Detection")
plt.axis("off")
plt.show()
```
